```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import random
```

```python
words = open('../data/names.txt','r').read().splitlines()
string = sorted(list(set((''.join(words)))))



stoi = { s: i + 1 for (i,s) in enumerate(string)}
itos = { i+1 : s for (i,s) in enumerate(string)}
stoi["."] = 0 
itos[0] = '.'
```

```python
block_size = 3

X, Y = [], []
random.seed(32)
random.shuffle(words)

n1 = int(0.8*len(words))
n2 = int(0.9*len(words))


def create_dataset(words) : 
    X, Y = [], []

    for w in words : 
        context = [0]*block_size

        for ch in w + '.' : 
            ix = stoi[ch]

            X.append(context)
            Y.append(ix)

            context = context[1:] + [ix]

    return (torch.tensor(X), torch.tensor(Y))

X_train, Y_train = create_dataset(words[:n1])
X_dev, Y_dev = create_dataset(words[n1:n2])
X_test, Y_test = create_dataset(words[n2:])


```

```python
#building parameters 
g = torch.Generator().manual_seed(234)

C = torch.randn((27,10),  generator = g )
W1 = torch.randn((30,200), generator=g )*0.1
b1= torch.randn(200, generator = g )*0.02
W2 = torch.randn((200, 27), generator = g)*0.1
b2 = torch.randn(27, generator=g)*0

parameters = [C, W1, b1, W2, b2]


for p in parameters : 
    p.requires_grad = True
```

```python
#training 


def train(X,Y) :    
    epochs = 200000
    for i in range(epochs) : 
        # mini batches
        ix = torch.randint(0, X_train.shape[0], (50,))
        emb = C[X[ix]]
        emb = torch.cat(torch.unbind(emb, dim = 1), dim=1)
        h = (emb@W1 + b1).tanh()
        logits = h@W2 + b2
        loss = F.cross_entropy(logits, Y_train[ix])

        #backward pass 
        for p in parameters : 
            p.grad = None

        loss.backward()

        lr = 0.1 if i < epochs * 0.6 else 0.01
        
        for p in parameters : 
            p.data += -lr*p.grad


    print(loss.item())


train(X_train,Y_train)
```

```python
#create some new words

g = torch.Generator().manual_seed(3455)

for i in range(20) : 

    out = []
    context = [0]*block_size

    while True : 
        emb = C[torch.tensor(context)]
        emb = emb.view(1, -1)
        h = (emb@W1 + b1).tanh()
        logits = h@W2 + b2
        probs = F.softmax(logits, dim=1)

        ix = torch.multinomial(probs, 1, generator=g).item()
        context = context[1:] + [ix]

        out.append(ix)
        if ix == 0 : 
            break

    print(''.join(itos[i] for i in out))


```
